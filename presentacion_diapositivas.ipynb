## 1. Problemática y Línea Base AS-IS (Jesús)
### Hallazgos de Campo del Laboratorio 01

- ⏱️ **Tiempo de atención prolongado:** Mediana de **0.83 horas (50 minutos)** por solicitud manual.
- 💰 **Alto costo unitario:** **S/ 3.66** de costo de mano de obra por cada cotización emitida.
- 📉 **Fricción en momentos de duelo:** Familiares expuestos a demoras innecesarias, llamadas repetitivas y falta de empatía estandarizada.
- ⚠️ **Riesgo de errores:** Cálculo manual de tarifas e IGV y falta de verificación de stock en tiempo real.

> **Objetivo TO-BE:** Reducir el tiempo a segundos mediante un sistema multi-agente que combine **rigurosidad técnica** con un **protocolo de condolencias y empatía**.

## 2. Arquitectura Lógica Multi-Agente BPMN 2.0 (Jesús)
### Desacoplamiento por Roles y Responsabilidades

```
           [ CLIENTE (Web / WhatsApp) ]
                       │ MF01: Solicitud digital
                       ▼
┌─────────────────────────────────────────────────────────────┐
│ LANE 1: ATENCIÓN AL CLIENTE (Agente Front-End)              │
│ • Recepción empática y validación de contrato de entrada    │
│ • Gestión de temporizadores de timeout (P2D)                │
└──────────────────────────────┬──────────────────────────────┘
                               ▼
┌─────────────────────────────────────────────────────────────┐
│ ORQUESTADOR DE FLUJO                                        │
│ • Coordinación secuencial de nodos F01 a F45 (BPMN TO-BE)   │
│ • Compuertas exclusivas y de eventos                        │
└──────────────────────────────┬──────────────────────────────┘
                               ▼
┌─────────────────────────────────────────────────────────────┐
│ LANE 2: ADMINISTRACIÓN (Agente Back-End)                    │
│ • Interrogación de Catálogo y Stock en Excel con pandas     │
│ • Idempotencia SHA-256, cálculo de IGV y Cola de Fallidos   │
└─────────────────────────────────────────────────────────────┘
```

## 3. Agente Front-End y Contrato de Entrada (Jhonny)
### Atención al Cliente y Protocolo ante el Duelo

- 🕊️ **Protocolo de Condolencias:** Respuesta inicial cálida y respetuosa ante la pérdida familiar.
- 📋 **Contrato de Entrada de Datos (Lab 02):**
  1. **Nombre del contacto:** Deudo o solicitante.
  2. **Teléfono celular:** Formato válido Perú (9 dígitos, inicia con 9).
  3. **Servicio funerario:** Sepelio, Cremación o Traslado.
  4. **Fecha requerida:** Formato estándar YYYY-MM-DD.
- ⏳ **Compuerta `F_GW_Datos` y Temporizador P2D:**
  - Si faltan datos: Solicita amablemente los faltantes.
  - Si no responde en 2 días (`P2D`): Dispara `F_TimeoutDatos` y archiva solicitud incompleta.

### Lane 2: Operaciones Automatizadas del Back-End

- ⚙️ **Orquestador de Flujo:** Implementa una máquina de estados garantizando la trazabilidad estricta del estándar BPMN 2.0.

## 6. Regla de Negocio: Alternativas y Tarifario (Franco)
### Capacidad Agotada y Desglose Transparente

- 🔄 **Nodo `F_DefinirAlternativa` (Compuerta `F_GW_Disponible`):**
  - Si un deudo solicita un servicio colmado (ej. *Cremación* sin capacidad en la fecha),
  - El Back-End detecta disponibilidad negativa y **asigna automáticamente una alternativa viable** (*Sepelio Integral Dignidad*), recalculando la cotización sin rechazar la venta.
- 💵 **Cálculo Económico Normativo:**
  - **Subtotal:** Precio base del servicio.
  - **IGV (18%):** Cálculo de ley exacto.
  - **Total (Soles / PEN):** Importe final transparente comunicado al deudo.

## 7. Robustez Técnica e Idempotencia SHA-256 (César)
### Seguridad Operativa en Entorno de Producción

- 🛡️ **Clave de Idempotencia SHA-256 (Prueba P2):**
  $$\text{Hash} = \text{SHA-256}(\text{Nombre} + \text{Teléfono} + \text{Servicio} + \text{Fecha})$$
  - Evita cotizaciones duplicadas si el cliente reenvía mensajes por WhatsApp o da doble clic.
- 📦 **Cola de Fallidos (Dead Letter Queue - `cola_errores.json`):**
  - Ante servicios inválidos o anomalías (ej. solicitud de *'AtaudPremium'*),
  - Se registra el incidente con **6 campos normativos** (ID, timestamp, error, clase, payload y reintentos) sin detener el servicio.
- 📂 **Persistencia:** Registro automático en `registro_cotizaciones_central.xlsx` (`DS_Central`).
